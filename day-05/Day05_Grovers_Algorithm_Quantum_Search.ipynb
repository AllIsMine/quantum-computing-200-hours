# Day 5 — Grover's Algorithm and Quantum Search

## Overview

This notebook demonstrates a two-qubit implementation of Grover's quantum search algorithm using Qiskit and Qiskit Aer.

Two target states are tested: `11` and `10`. Each experiment uses an oracle to mark the target state, a diffusion operator to amplify its amplitude, and measurement to check the result.

## Learning Objectives

- Prepare a two-qubit superposition.
- Implement and modify a Grover oracle.
- Apply the diffusion operator.
- Simulate quantum circuits using Qiskit Aer.
- Validate measurement results using Python assertions.
- Understand qubit indexing, measurement output, and simulator limitations.

## Experiments

1. Search for target state `11`.
2. Modify the oracle and search for target state `10`.

**Environment:** Python, Qiskit, Qiskit Aer, and Google Colab.

**Note:** Results in this notebook are from ideal simulation, not real quantum hardware.

````python
%pip install qiskit qiskit-aer
````

````python
from qiskit import QuantumCircuit

# Create a two-qubit circuit
qc = QuantumCircuit(2, 2)

# Put both qubits into superposition
qc.h([0, 1])

# Display the circuit
print(qc.draw())
````

````python
# Mark the target state |11>
qc.cz(0, 1)

# Display the circuit
print(qc.draw())
````

````python
# Diffusion operator
qc.h([0, 1])
qc.x([0, 1])
qc.cz(0, 1)
qc.x([0, 1])
qc.h([0, 1])

# Display the circuit
print(qc.draw())
````

````python
# Measure both qubits
qc.measure([0, 1], [0, 1])

# Display the complete circuit
print(qc.draw())
````

````python
from qiskit_aer import AerSimulator

# Run the circuit 1000 times
simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()

# Display the measurement results
counts = result.get_counts()
print("Measurement counts:", counts)
````

````python
assert counts == {"11": 1000}
print("Day 5 Grover's algorithm test passed!")
````

````python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc10 = QuantumCircuit(2, 2)

# 1. Prepare superposition
qc10.h([0, 1])

# 2. Oracle for target |10>
qc10.x(0)
qc10.cz(0, 1)
qc10.x(0)

# 3. Diffusion operator
qc10.h([0, 1])
qc10.x([0, 1])
qc10.cz(0, 1)
qc10.x([0, 1])
qc10.h([0, 1])

# 4. Measure
qc10.measure([0, 1], [0, 1])

# 5. Simulate
simulator = AerSimulator()
result10 = simulator.run(qc10, shots=1000).result()
counts10 = result10.get_counts()

print("Measurement counts:", counts10)
assert counts10 == {"10": 1000}
print("Grover target-10 test passed!")
````

## Experiment 2 — Searching for a Different Target

### Objective

Test whether Grover's algorithm can find a different target state by modifying the oracle while keeping the diffusion operator unchanged.

### Background

Grover's algorithm searches for a marked state in an unsorted search space. It uses quantum superposition, an oracle, and amplitude amplification to increase the probability of measuring the target state.

### Method

The first experiment searched for the target state `11`. In this experiment, the target is changed to `10`.

The circuit follows these steps:

1. **Superposition:** Apply Hadamard gates to both qubits to create an equal superposition of the four computational basis states.
2. **Oracle:** Apply an X gate to qubit 0, followed by a controlled-Z gate, and then another X gate on qubit 0. This marks the desired state `10` with a phase change.
3. **Diffusion:** Apply the diffusion operator to amplify the target state's amplitude.
4. **Measurement:** Measure both qubits.
5. **Simulation:** Run the circuit using Qiskit Aer with 1,000 shots.

### Result

The ideal Qiskit Aer simulation produced:

```text
Measurement counts: {'10': 1000}
Grover target-10 test passed!
```

### Interpretation

The target state `10` appeared in all 1,000 simulator shots. The assertion passed, confirming that the modified circuit produced the expected result in this ideal simulation.

### Debugging Lesson

An initial version of the oracle produced `01` instead of the intended `10`. Reviewing Qiskit's qubit indexing and classical bit-string ordering helped identify the error. Changing the X-gate operations to act on qubit 0 corrected the result.

This debugging process reinforced the importance of understanding how qubit indices map to displayed measurement outcomes.

### Limitation

These results were obtained using an ideal simulator, not real quantum hardware. Actual hardware results can vary because of noise, gate errors, and measurement errors. This two-qubit example demonstrates the algorithm's basic mechanism; it does not establish a practical speed advantage for large search problems.

### Conclusion

This experiment demonstrates how changing the oracle changes the target state found by Grover's algorithm. Together, the `11` and `10` experiments provide practical experience with oracle construction, diffusion, measurement, simulation, assertions, and debugging.

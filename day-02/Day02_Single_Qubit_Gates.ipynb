```python

import qiskit
from qiskit.quantum_info import Statevector

print("Qiskit version:", qiskit.__version__)
print("Statevector simulator is ready!")
```

```python

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a circuit with one qubit
qc = QuantumCircuit(1)

# Apply the Hadamard gate
qc.h(0)

# Calculate the quantum state
state = Statevector.from_instruction(qc)

# Display the circuit and results
print(qc.draw())
print("Statevector:", state)
print("Measurement probabilities:", state.probabilities_dict())
```

```python

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a single-qubit circuit
qc = QuantumCircuit(1)

# Apply the Pauli-X gate
qc.x(0)

# Calculate the final quantum state
state = Statevector.from_instruction(qc)

print(qc.draw())
print("Statevector:", state)
print("Probabilities:", state.probabilities_dict())
```

```python

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Prepare a qubit in |0>
qc = QuantumCircuit(1)

# Create superposition, then change its phase
qc.h(0)
qc.z(0)

# Inspect the final state
state = Statevector.from_instruction(qc)

print(qc.draw())
print("Statevector:", state)
print("Probabilities:", state.probabilities_dict())
```

```python

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

qc.h(0)
qc.z(0)
qc.h(0)

state = Statevector.from_instruction(qc)

print(qc.draw())
print("Final state:", state)
print("Probabilities:", state.probabilities_dict())
```

```python

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

qc.x(0)
qc.x(0)

state = Statevector.from_instruction(qc)

print(qc.draw())
print("Final state:", state)
print("Probabilities:", state.probabilities_dict())

assert state.probabilities()[0] == 1.0
print("Test passed: Two X gates return the qubit to |0>.")
```

## Day 2 Summary — Single-Qubit Gates

### Concepts learned
- Qubit statevectors and probability amplitudes
- Superposition using the Hadamard gate
- Bit flips using the Pauli-X gate
- Phase changes using the Pauli-Z gate
- How phase affects later quantum operations
- Reversibility of quantum gates

### Experiments completed
1. Hadamard gate applied to |0⟩.
2. Pauli-X gate applied to |0⟩.
3. H → Z → H circuit.
4. X → X circuit.

### Key observations
- H|0⟩ creates an equal superposition.
- X|0⟩ = |1⟩.
- H → Z → H transforms |0⟩ into |1⟩.
- Applying X twice returns |0⟩ to |0⟩.

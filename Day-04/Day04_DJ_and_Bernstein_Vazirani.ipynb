```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit (1,1)

qc.h(0)

qc.measure(0,0)

simulator = AerSimulator()

result = simulator.run(qc,shots=1000).result()

counts = result.get_counts()

print(counts)
```

```python
qc = QuantumCircuit(4)

# Prepare the ancilla
qc.x(3)

# Create superposition
qc.h([0, 1, 2, 3])

# Balanced oracle
qc.cx(0, 3)

print(qc.draw())
```

```python
qc = QuantumCircuit(4)

# Prepare ancilla and superposition
qc.x(3)
qc.h([0, 1, 2, 3])

# Balanced oracle
qc.cx(0, 3)

# Final Hadamard gates on input qubits
qc.h([0, 1, 2])

print(qc.draw())
```

```python
qc = QuantumCircuit(4, 3)

qc.x(3)
qc.h([0, 1, 2, 3])
qc.cx(0, 3)
qc.h([0, 1, 2])

qc.measure([0, 1, 2], [0, 1, 2])

print(qc.draw())
```

```python
qc = QuantumCircuit(4, 3)

qc.x(3)
qc.h([0, 1, 2, 3])
qc.cx(0, 3)
qc.h([0, 1, 2])

qc.measure([0, 1, 2], [0, 1, 2])

simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()
counts = result.get_counts()

print(counts)
```

```python
qc = QuantumCircuit(4, 3)

qc.x(3)
qc.h([0, 1, 2, 3])

# Constant-zero oracle: no CNOT gate

qc.h([0, 1, 2])
qc.measure([0, 1, 2], [0, 1, 2])

simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()

print(result.get_counts())
```

```python
qc = QuantumCircuit(4, 3)

qc.x(3)
qc.h([0, 1, 2, 3])

# Constant-one oracle
qc.x(3)

qc.h([0, 1, 2])
qc.measure([0, 1, 2], [0, 1, 2])

simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()

print(result.get_counts())
```

```python
assert result.get_counts() == {'000': 1000}
print("Constant oracle test passed!")
```

```python
assert len(counts) == 1
assert '000' not in counts

print("Balanced oracle test passed!")
```

```python
from qiskit import QuantumCircuit

secret = "101"
n = len(secret)

qc = QuantumCircuit(n + 1)

for i, bit in enumerate(reversed(secret)):
    if bit == "1":
        qc.cx(i, n)

print(qc.draw())
```

```python
from qiskit import QuantumCircuit

secret = "101"
n = len(secret)

qc = QuantumCircuit(n + 1)

# Prepare the ancilla qubit
qc.x(n)

# Put all qubits into superposition
qc.h(range(n + 1))

print(qc.draw())
```

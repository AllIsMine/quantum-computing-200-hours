```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

n = 3
qc = QuantumCircuit(n + 1, n)

# Prepare the ancilla in |1>
qc.x(n)

# Create superposition
qc.h(range(n + 1))

# Constant-zero oracle: no oracle gates needed

# Apply final Hadamards to input qubits
qc.h(range(n))

# Measure input qubits
qc.measure(range(n), range(n))

# Simulate
simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()
counts = result.get_counts()

print("Constant-zero counts:", counts)

assert counts == {"000": 1000}
print("Constant-zero test passed!")
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

n = 3
qc = QuantumCircuit(n + 1, n)

# Prepare ancilla
qc.x(n)

# Create superposition
qc.h(range(n + 1))

# Balanced oracle: f(x) = x[0]
qc.cx(0, n)

# Interference
qc.h(range(n))

# Measure input register
qc.measure(range(n), range(n))

# Run actual circuit
result = AerSimulator().run(qc, shots=1000).result()
counts = result.get_counts()

print("Actual balanced counts:", counts)

# Validate the actual result
assert len(counts) == 1
assert "000" not in counts

print("Actual balanced-circuit validation passed!")
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Hidden secret
secret = "101"
n = len(secret)

# Three input qubits + one ancilla
qc = QuantumCircuit(n + 1, n)

# 1. Prepare ancilla in |1>
qc.x(n)

# 2. Create superposition
qc.h(range(n + 1))

# 3. Encode secret in the oracle
for i, bit in enumerate(reversed(secret)):
    if bit == "1":
        qc.cx(i, n)

# 4. Apply final Hadamard gates
qc.h(range(n))

# 5. Measure input qubits
qc.measure(range(n), range(n))

# 6. Run simulator
result = AerSimulator().run(qc, shots=1000).result()
counts = result.get_counts()

print("Bernstein–Vazirani counts:", counts)

# 7. Verify the recovered secret
assert counts == {"101": 1000}
print("Bernstein–Vazirani test passed!")
```

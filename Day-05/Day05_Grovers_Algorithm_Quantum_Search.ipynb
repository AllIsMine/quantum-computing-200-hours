```python
from qiskit import QuantumCircuit

# Create a two-qubit circuit
qc = QuantumCircuit(2, 2)

# Put both qubits into superposition
qc.h([0, 1])

# Display the circuit
print(qc.draw())
```

```python
# Mark the target state |11>
qc.cz(0, 1)

# Display the circuit
print(qc.draw())
```

```python
# Diffusion operator
qc.h([0, 1])
qc.x([0, 1])
qc.cz(0, 1)
qc.x([0, 1])
qc.h([0, 1])

# Display the circuit
print(qc.draw())
```

```python
# Measure both qubits
qc.measure([0, 1], [0, 1])

# Display the complete circuit
print(qc.draw())
```

```python
from qiskit_aer import AerSimulator

# Run the circuit 1000 times
simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()

# Display the measurement results
counts = result.get_counts()
print("Measurement counts:", counts)
```

```python
assert counts == {"11": 1000}
print("Day 5 Grover's algorithm test passed!")
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc10 = QuantumCircuit(2, 2)

# 1. Prepare superposition
qc10.h([0, 1])

# 2. Oracle for target |10>
qc10.x(0)
qc10.cz(0, 1)
qc10.x(0)

# 3. Diffusion operator
qc10.h([0, 1])
qc10.x([0, 1])
qc10.cz(0, 1)
qc10.x([0, 1])
qc10.h([0, 1])

# 4. Measure
qc10.measure([0, 1], [0, 1])

# 5. Simulate
simulator = AerSimulator()
result10 = simulator.run(qc10, shots=1000).result()
counts10 = result10.get_counts()

print("Measurement counts:", counts10)
assert counts10 == {"10": 1000}
print("Grover target-10 test passed!")
```

## Experiment 2 — Searching for a Different Target

### Objective
Test whether the Grover circuit can find a different target state by changing the oracle.

### Method
The original circuit searched for `11`. For the second experiment, the oracle was modified to mark `10` by applying an X gate to qubit 0 before and after the controlled-Z gate.

### Result

The Qiskit Aer simulation was run for 1,000 shots.

\`\`\`text
Measurement counts: {'10': 1000}
Grover target-10 test passed!
\`\`\`

### Interpretation
The simulation returned the intended target state `10` in all 1,000 shots. This confirms that the modified oracle and diffusion circuit work as expected for this target in the ideal simulator.

### Debugging Lesson
The initial oracle modification produced `01` rather than `10`. Reviewing Qiskit's qubit indexing and displayed measurement-bit order helped identify and correct the issue.

### Limitation
These results come from an ideal simulator. Real quantum hardware can produce different results because of noise and gate errors.
